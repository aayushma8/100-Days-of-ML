This code extracts the prices from the table. First, price = [] creates an empty list to store the prices. The first for loop goes through each <tbody>, and the second for loop goes through each <tr> (table row/company). row.find_all('td') gets all the cells in that row and stores them in td. We use td[4] because the Price is in the 5th <td> of the row, and Python starts counting from 0, so the 5th item has index 4. if len(td) > 4 checks that the 5th <td> actually exists before accessing it. Finally, .text.strip() gets and cleans the price text, and append() adds it to the price list.

```python

price = []

for i in company:
    for row in i.find_all('tr'):
        td = row.find_all('td')
        if len(td) > 4:
            price.append(td[4].text.strip())

```

```python
earnings = []

for row in company[0].find_all('tr'):
    td = row.find_all('td')
    if len(td) > 3:
        earnings.append(td[3].text.strip())

print(len(earnings))
```

We first stored all the <tbody> elements found on the webpage in the company variable using soup.find_all('tbody'). Since there can be multiple <tbody> elements, company[0] selects the first <tbody> from the list. We then use .find_all('tr') to find all the table rows inside that first <tbody>. Since .find_all('tr') returns a list of rows, [0] selects the first row. Finally, .text extracts the text from that row, and .strip() removes any extra spaces and line breaks.
