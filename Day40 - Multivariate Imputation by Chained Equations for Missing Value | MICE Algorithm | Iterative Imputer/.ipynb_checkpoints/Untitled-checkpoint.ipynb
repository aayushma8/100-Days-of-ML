```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
```

```python
df = np.round(pd.read_csv('50_Startups.csv')[['R&D Spend','Administration','Marketing Spend','Profit']]/10000)

np.random.seed(9) # Seed = control randomness so it gives the same result every time.

df=df.sample(5)

df
```

```python
#remove the target column
df = df.iloc[:, 0:-1]
df
```

```python
#  introducted some fake NaN Values

df.iloc[1,0] = np.nan # rows, column
df.iloc[3,1] = np.nan
df.iloc[-1,-1] = np.nan
```

```python
df.head()
```

```python
# step 1 - Impute all missing values with mean of respective col

df0 = pd.DataFrame()

df0['R&D Spend'] = df['R&D Spend'].fillna(df['R&D Spend'].mean())
df0['Administration'] = df['Administration'].fillna(df['Administration'].mean())
df0['Marketing Spend'] = df['Marketing Spend'].fillna(df['Marketing Spend'].mean())
```

```python
# 0th iteration
df0
```

```python
# Step 2 : Remove Col 1  imputed value

df1 = df0.copy()

df1.iloc[1,0] = np.nan

df1
```

```python
# use all other rows except for row 1  to build a model
X= df1.iloc[[0,2,3,4], 1:3] # 1:3 means column 1 and 2
X
```

```python
y = df1.iloc[[0, 2, 3, 4],0] # rows, columns
y # we are making this y
```

```python
# Step 3 predict missing value of cols 1 using other cols
lr = LinearRegression()

# train the model
lr.fit(X,y)

# Predict for row 1
lr.predict(df1.iloc[[1], 1:]) # missing row to data act as i/p for prediction

lr.predict(df1.iloc[1,1:].values.reshape(1,2)) 
```

```python
df1.iloc[1,0] = 23.14
df1
```

```python
# step 4 Remove all cols imputedd value

df1.iloc[3,1] = np.nan

df1
```

```python
# step 5: Predict the missing value of cols2 using other columns
X=df1.iloc[[0,1,2,4],[0,2]]
X
```

```python
y = df1.iloc[[0,1,2,4],1]
y
```

```python
lr = LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[[3], [0, 2]])
```

```python
df1.iloc[3,1] = 11.06
```

```python
# step 7: Predict the missing value of cols3 using other columns

X= df1.iloc[0:4, 0:2]
X
```

```python
y = df1.iloc[0:4, 2]
y
```

```python
lr = LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[[4], [0, 1]])  # missing row to data act as i/p for prediction
```
