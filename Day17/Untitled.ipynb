```python
import pandas as pd
import requests

# 1. Create a blank list to temporarily store DataFrames from each API page
temp_list = []

# 2. Loop variable 'i' acts as our page counter, running from 1 to 500
for i in range(1, 501):

    # 3. .format(i) replaces '{}' right after 'page=' with number 'i', dynamically fetching page=1, page=2, etc.
    response = requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=fe20fc7fd2c11faba63d5c4c310ec922&language=en-US&page={}'.format(i))

    # 4. Convert TMDb's JSON response string into a Python dictionary object
    data = response.json()

    # 5. Extract the array from data['results'] and build a DataFrame with only specified columns
    temp_df = pd.DataFrame(data['results'])[['id', 'title', 'overview', 'release_date', 'popularity', 'vote_average', 'vote_count']]

    # 6. Add the current page's temporary DataFrame into our list
    temp_list.append(temp_df)

# 7. Merge all 500 DataFrames in temp_list into one master DataFrame
# ignore_index=True re-indexes all rows continuously from 0 to 9999 instead of resetting to 0 for every page 
# and creates a new continuous row order (0 to 9999) for the final table.
df = pd.concat(temp_list, ignore_index=True)
```
