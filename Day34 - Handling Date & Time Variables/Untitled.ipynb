# Woking with Dates

```python
# converting to datetime datatype

# dataframe_name['column_name'] = pd.to_datetime(dataframe_name['column_name'])

date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

### 1. Extract year

```python
# dataframe_name['new_year_column'] = dataframe_name['date_column_name'].dt.year
date['date_year'] = date['date'].dt.year
date.sample(5)
```

### 2. Extract Month

```python
date['date_month_no'] = date['date'].dt.month
date.head()
```

```python
date['date_month_name'] = date['date'].dt.month_name()
date.head()
```

### Extract Days

```python
date['date_day'] = date['date'].dt.day
date.head()
```

### Day of Week

```python
date['date_dow'] = date['date'].dt.dayofweek
date.sample(3)
```

### Day of Week - name

```python
date['date_dow_name'] = date['date'].dt.day_name()

date.drop(columns=['product_id', 'city_id', 'orders']).head()
```

### Is Weekend?

```python
#dataframe_name['new_column_name'] = np.where(dataframe_name['column_name'].isin(['Match1', 'Match2']), 'Value if True', 'Value if False')

date['date_is _weekend'] = np.where(date['date_dow_name'].isin(['Sunday', 'Saturday']), 1, 0)

date.drop(columns=['product_id', 'city_id', 'orders']).head()
```

### Extract week of the year

```python
date['date_week'] = date['date'].dt.isocalendar().week

date.drop(columns=['product_id', 'city_id', 'orders']).head()
```

### Extract Quarter

```python
date['quarter'] = date['date'].dt.quarter

date.drop(columns=['product_id', 'city_id', 'orders']).head()
```

### Extract Semester

```python
date['Semester'] = np.where(date['quarter'].isin([1,2]), 1, 2)

date.drop(columns=['product_id', 'city_id', 'orders']).head()
```

### Extract Time Elapsed Between Dates

```python
import datetime

today = datetime.datetime.today()

today
```

```python
today - date['date']
```

```python
(today - date['date']).dt.days
```

```python
# months passed

np.round((today - date['date']).dt.days / 30.44, 0)
```
