```python
import numpy as np
import pandas as pd

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression

import seaborn as sns
```

```python
df = pd.read_csv('train.csv')[['Age', 'Pclass', 'SibSp', 'Parch', 'Survived']]
```

```python
X= df.iloc[:, 0:4]
y = df.iloc[:, -1]
```

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(), X, y, scoring='accuracy', cv=20))
```

# Applying Feature Construction

```python
'''
SibSp: Number of Siblings + Spouses traveling with the passenger.

Parch: Number of Parents + Children traveling with the passenger.

+1: Represents the passenger.'''




X['Family_size'] = X['SibSp'] + X['Parch'] + 1 
```

```python
X.head()
```

```python
def myfunc(num):
    if num == 1: # if the passenger is travelling alone
        return 0
    elif num>1 and num<=4: # small family
        return 1
    else: # large family
        return 2
```

```python
myfunc(4)
```

```python
myfunc(888)
```

```python
X['Family_type'] = X['Family_size'].apply(myfunc)
```

```python
X.head()
```

```python
X.drop(columns=['SibSp', 'Parch', 'Family_size'], inplace = True)
```

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(), X, y, scoring='accuracy', cv=20))
```
