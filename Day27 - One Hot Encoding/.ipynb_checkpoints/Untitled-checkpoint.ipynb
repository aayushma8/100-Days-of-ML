# 1. OneHotEncoding using Pandas

```python
# here we didn't solve multicolinearity problem i.e. n(columns)-1
pd.get_dummies(df, columns=['fuel', 'owner'])
```

# 2. K-1 OneHotEncoding

```python
# we dont normally use these first 2 OHE technique in ML projects
pd.get_dummies(df, columns=['fuel', 'owner'], drop_first=True)
# drop_first=True means remove the first dummy column for each categorical variable.
```

# 3. OneHotEncoding using Sklearn

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 0:4], df.iloc[:, -1], test_size=0.2, random_state=2)
```

```python
X_train.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
# Create an object from a class OneHotEncoder
#sparse_output=False → normal array
#dtype=np.int32 → integers (0 and 1)
#pandas  → drop_first=True and sklearn → drop='first'
# drop = 'first' For each categorical column, remove the first category's dummy column.
ohe = OneHotEncoder(drop='first', sparse_output=False, dtype=np.int32)
```

### We leave 'brand' and 'km_driven' for now: 'km_driven' is numerical, while 'brand' has many categories and would create too many columns with OHE.

```python
ohe.fit_transform(X_train[['fuel', 'owner']]) # did fit and transform together for X_train data columns ie fuel and ow
```

```python
X_train_new = ohe.fit_transform(X_train[['fuel', 'owner']])
```

```python
# remember we always fit train data only not test data
X_test_new = ohe.transform(X_test[['fuel', 'owner']])
```
