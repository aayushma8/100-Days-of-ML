```python
import pickle
import numpy as np
```

```python
# DECODING# Load the previously saved encoders and model from the models folder
# 'rb' means read in binary mode

ohe_sex = pickle.load(open('models/ohe_sex.pkl', 'rb'))
ohe_embarked = pickle.load(open('models/ohe_embarked.pkl', 'rb'))
clf = pickle.load(open('models/clf.pkl', 'rb'))
```

```python
# Assume user input
# Pclass/gender/age/SibSp/Parch/Fare/Embarked
 # Create a new passenger's data to test our trained model and predict whether they survived or not.

test_input = np.array([2, 'male', 31.0, 0, 0, 10.5, 'S'], dtype=object).reshape(1,7)
```

```python
test_input
```

```python
# Extract the new passenger's Sex and encode it using the trained OneHotEncoder.
test_input_sex = ohe_sex.transform(test_input[:, 1].reshape(1, 1))
```

```python
test_input_sex # it is male so the o/p will be 0,1
```

```python
# Extract the new passenger's Embarked value and encode it using the trained Embarked encoder.
test_input_embarked = ohe_embarked.transform(test_input[:, -1].reshape(1, 1))
```

```python
test_input_embarked # for output S 001 is the code 
```

```python
test_input_age = test_input[:, 2].reshape(1, 1)
```

```python
test_input_age
```

```python
test_input_transformed = np.concatenate((test_input[:, [0, 3, 4, 5]], test_input_age, test_input_sex, test_input_embarked), axis = 1)
# Take all rows (:), but only take columns 0, 3, 4, and 5.
```

```python
test_input_transformed
```

```python
test_input_transformed.shape
```

```python
clf.predict(test_input_transformed) #The model predicts that this passenger survived i.e. ou put is (1)
```
